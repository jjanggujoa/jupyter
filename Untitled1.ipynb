```python
import requests
```

```python
url = 'http://openapi.seoul.go.kr:8088/625a43554c73656f37344e496d6757/json/viewNightSpot/1/100/'
response = requests.get(url)
```

```python
print(response.content)
```

```python
data=response.json()
data
```

```python
import pandas as pd
temp=[]
for d in data['viewNightSpot']['row']:
    temp.append(d)
df=pd.DataFrame(temp)
df
```

```python
import folium

folium.Map()
```

```python
night=folium.Map(location=[37.5665, 126.9780], zoom_start=12)
night
```

```python
night=folium.Map(zoom_start=12)
night
```

```python
night=folium.Map(location=[37.5665, 126.9780], zoom_start=12,zoom_control=False,control_scale=True)
night
```

```python
for n, row in df.iterrows():
    folium.Marker([row['LA'], row['LO']], popup=row['ADDR']).add_to(night)
night
```

```python
night2= folium.Map(location=[37.5665, 126.9780], zoom_start=12,zoom_control=False,control_scale=True)
night2
```

```python
from folium.plugins import MarkerCluster
mc=MarkerCluster().add_to(night2)

for n, row in df.iterrows():
    folium.Marker([row['LA'], row['LO']], popup=row['ADDR']).add_to(mc)
night2
```
