```python
import pandas as pd

df=pd.read_excel('/content/drive/MyDrive/자연휴양림출력화일.xlsx')
df.head()
```

```python
anMap=folium.Map(location=[df['위도'].mean(),df['경도'].mean()],zoom_start=10,control_scale=True,zoom_control=False)
anMap
```

```python
for n, row in df.iterrows():
  print(n, row['휴양림명'],row['위도'],row['경도'])
```

```python
for n, row in df.iterrows():
  mk=folium.Marker(location=[row['위도'],row['경도']],popup=row['휴양림명'])
  mk.add_to(anMap)
anMap
```

```python
anMap2=folium.Map(location=[37.4965,126.9573],zoom_start=10,control_scale=True,zoom_control=False)
anMap2
```

```python
from folium.plugins import MarkerCluster

mc=MarkerCluster().add_to(anMap2)
for n, row in df.iterrows():
  mk=folium.Marker(location=[row['위도'],row['경도']],popup=row['휴양림명'],max_width=200).add_to(mc)
anMap2
```
